# Value at Risk (VaR)

Value at Risk (VaR) is a widely used risk management tool that quantifies the potential loss in value of a portfolio over a specified time period for a given confidence level. It provides an estimate of the maximum loss that an investor can expect to incur with a certain probability.

## Key Concepts
- **Confidence Level**: The probability that the actual loss will not exceed the VaR estimate. Common confidence levels are 95% and 99%.
- **Time Horizon**: The period over which the VaR is calculated, such as one day, one week, or one month.
- **Loss Distribution**: The statistical distribution of potential losses, which can be estimated using historical data or assumed to follow a specific distribution (e.g., normal distribution).

## Methods to Calculate VaR
1. **Historical Simulation**: This method uses historical returns to simulate potential losses. It involves sorting the historical returns and selecting the appropriate percentile based on the confidence level.
2. **Variance-Covariance Method**: This method assumes that returns are normally distributed and calculates VaR using the mean and standard deviation of the returns. The formula is:
$$\text{VaR} = \mu + z \cdot \sigma $$
   where $ \mu $ is the mean return, $ z $ is the z-score corresponding to the confidence level, and $ \sigma $ is the standard deviation of returns.

   
3. **Monte Carlo Simulation**: This method involves generating a large number of random scenarios for portfolio returns based on assumed distributions and calculating the potential losses for each scenario to estimate VaR.

## Limitations of VaR
- **Assumption of Normality**: Many VaR models assume that returns are normally distributed, which may not always be the case, especially during periods of market stress.
- **Does Not Capture Tail Risk**: VaR does not provide information about losses that exceed the VaR threshold, which can be significant in extreme market conditions.
- **Static Measure**: VaR is a snapshot of risk at a specific point in time and may not reflect changes in market conditions or portfolio composition.
- **Model Risk**: The accuracy of VaR estimates depends on the chosen model and assumptions, which can lead to underestimation or overestimation of risk if the model is not appropriate for the data.

# Expected Shortfall (ES)

Expected Shortfall (ES), also known as Conditional Value at Risk (CVaR), is a risk measure that provides an estimate of the average loss given that the loss exceeds the VaR threshold. It is considered a more comprehensive measure of risk than VaR, as it captures tail risk and provides information about potential losses in extreme market conditions.

The formula for Expected Shortfall is:
$$\text{ES} = E[L | L > \text{VaR}]$$
where $ L $ is the loss and $ \text{VaR} $ is the Value at Risk threshold.

Expected Shortfall is particularly useful for risk management as it provides insights into the severity of losses that can occur beyond the VaR threshold, making it a valuable tool for investors and financial institutions to assess and manage their risk exposure effectively.

### Why should we use mean instead of median for expected shortfall? 

If we use median, we would be underestimating the expected shortfall, we are ignoring the extreme losses that occur beyond the VaR threshold ("fat tails"). The mean provides a more accurate representation of the average loss in extreme market conditions, while the median may give a skewed view of the potential losses.

```python
import yfinance as yf  
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# VaR using Historical Method

```python
data = yf.download('SPY', start='2010-01-01', end='2023-01-01', progress=False)['Close']
returns = data.pct_change().dropna()

condfidence_level = 0.95
var_95 = np.percentile(returns, (1 - condfidence_level) * 100)
print(f"Value at Risk (VaR) at {condfidence_level*100}% confidence level: {var_95:.4f}")

worst_days = returns['SPY'][returns['SPY'] < var_95]
expected_shortfall_95 = worst_days.mean()
print(f"Expected Shortfall (ES) at {condfidence_level*100}% confidence level: {expected_shortfall_95:.4f}") 
```
