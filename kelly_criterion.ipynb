# Kelly Criterion

Kelly Criterion is a formula used to determine the optimal size of a series of bets in order to maximize the logarithm of wealth. It is commonly used in gambling and investing to manage risk and maximize returns. The formula is given by:
$$f^* = p - \frac{q}{b}$$

Where:
- $f^*$ is the fraction of the current wealth to bet
- $b$ is the net odds received on the bet (i.e., the amount won per unit bet)
- $p$ is the probability of winning
- $q$ is the probability of losing $(q = 1 - p)$

In the context of investing, the Kelly Criterion can be used to determine how much of a portfolio to allocate to a particular asset or strategy based on the expected return and the probability of success. It helps investors to avoid overbetting and underbetting, thus optimizing their long-term growth.
