For options, black scholes formula is

$$c = S_0 N(d_1) - K e^{-rt} N(d_2)$$
$$p = K e^{-rt} N(-d_2) - S_0 N(-d_1)$$
$$d_1 = \frac{\ln(S_0/K) + (r + \sigma^2/2)t}{\sigma \sqrt{t}}$$
$$d_2 = d_1 - \sigma \sqrt{t}$$

Where:
- $c$ is the call option price
- $S_0$ is the current stock price
- $K$ is the strike price
- $r$ is the risk-free interest rate
- $t$ is the time to maturity
- $\sigma$ is the volatility of the stock
- $N(d)$ is the cumulative distribution function of the standard normal distribution evaluated at $d$


```python
import numpy as np
from scipy.stats import norm

S_0 = 49
K = 50
r = 0.05
sigma = 0.2
T = 0.3846 # 20 weeks
mu = 0.13

d1 = (np.log(S_0 / K) + (r + sigma**2/2) * T) / (sigma * np.sqrt(T)) 
d2 = d1 - sigma * np.sqrt(T)

c = S_0 * norm.cdf(d1) - K * np.exp(- r * T) * norm.cdf(d2)
p = K * np.exp(- r * T) * norm.cdf(-d2) - S_0 * norm.cdf(-d1) 

print(c, p)
```
