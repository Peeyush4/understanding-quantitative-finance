# Implied Volatility



```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt 
import numpy as np
import scipy.stats as stats
```

```python
def bs_call_price(S, K, T, r, sigma):
    d1 = (np.log(S / K) + (r + 0.5 * sigma ** 2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    return S * stats.norm.cdf(d1) - K * np.exp(-r * T) * stats.norm.cdf(d2)
    
# Vega = df/dT
def bs_vega(S, K, T, r, sigma):
    d1 = (np.log(S / K) + (r + 0.5 * sigma ** 2) * T) / (sigma * np.sqrt(T))
    return S * stats.norm.pdf(d1) * np.sqrt(T)

def implied_volatility(S, K, T, r, market_price, threshold=1e-5, max_iterations=100):
    # This is a simple implementation using the Newton-Raphson method
    sigma = 0.2  # Initial guess
    for i in range(max_iterations):
        bs_price = bs_call_price(S, K, T, r, sigma)
        vega = bs_vega(S, K, T, r, sigma)
        if abs(bs_price - market_price) < threshold:
            return sigma
        sigma -= (bs_price - market_price) / vega
    return np.nan
```

```python
target_market_price = 10.45

iv = implied_volatility(S=100, K=100, T=1.0, r=0.05, market_price=target_market_price)
print(f"The Implied Volatility is: {iv:.2%}")
```
